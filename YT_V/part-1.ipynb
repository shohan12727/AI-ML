```python
import numpy as np
```

```python
a = np.array([1,2,3])
b = np.array([2,4,8])
a + b
```

```python
# 1D array
arr1 = np.array([1, 2, 3, 4, 5])

# 2D array
arr2 = np.array([[1, 2, 3], [4, 5, 6]])

# 3D array 
arr3 = np.array([
    [[1, 2, 3], [4, 5, 6]],
    [[1, 2, 3], [4, 5, 6]],
    [[1, 2, 3], [4, 5, 6]]
])
```

```python
print(arr1.ndim)
print(arr2.ndim)
print(arr3.ndim)
```

```python
print(arr1.shape)
print(arr2.shape)
print(arr3.shape)
```

```python
print(arr1.dtype)
print(arr2.dtype)
print(arr3.dtype)
```

```python
print(arr1.size)
print(arr2.size)
print(arr3.size)
```

```python
arr = np.array([1,2,3])
print(arr.dtype)

arr = np.array([1.3,5.6,8.5])
print(arr.dtype)
```

```python
arr = np.array([1,2,4,5,6.8,3.6])
print(arr.dtype)

```

```python
arr = np.array([1,2,3.5,"Shohan"])
print(arr.dtype)
```

```python
arr = np.array([1,2,300], dtype=np.int16)
changed_the_data_type = arr.dtype
print(changed_the_data_type)
```

```python
arr = np.array([1,2.5,300.96])
print(arr.dtype)
modified_array = arr.astype(np.int32)
print(modified_array.dtype)
```

```python
arr = np.array([1,2,4.5,"shohan","car"], dtype = np.float64)
print(arr.dtype)
```
